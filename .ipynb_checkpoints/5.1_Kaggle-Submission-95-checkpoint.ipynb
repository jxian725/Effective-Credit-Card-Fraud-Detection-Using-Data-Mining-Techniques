```python
import pandas as pd
import numpy as np
import lightgbm as lgb
from sklearn.preprocessing import StandardScaler
from category_encoders import TargetEncoder
import os

# -------------------- Paths --------------------
train_path = "datasets/train_dataset.csv"
test_transaction_path = "datasets/test_transaction.csv"
test_identity_path = "datasets/test_identity.csv"
output_dir = "IEEE-CIS_Submission"
model_path = "models/lightgbm_95.txt"
selected_feature_path = "feature_importance_reduced/lgbm_feature_importance_target_encoded_1.csv"
submission_path = f"{output_dir}/1_lightgbm-submission_95-dataset.csv"

os.makedirs(output_dir, exist_ok=True)
os.makedirs("models", exist_ok=True)

# -------------------- Load Datasets --------------------
train_data = pd.read_csv(train_path)
X = train_data.drop(columns=["isFraud"])
y = train_data["isFraud"]

# Load and normalize test data
test_transaction = pd.read_csv(test_transaction_path)
test_identity = pd.read_csv(test_identity_path)

test_transaction.columns = test_transaction.columns.str.replace('-', '_').str.strip()
test_identity.columns = test_identity.columns.str.replace('-', '_').str.strip()

test_data = test_transaction.merge(test_identity, how='left', on='TransactionID')
submission = pd.DataFrame()
submission["TransactionID"] = test_data["TransactionID"]

# -------------------- Feature Selection --------------------
selected_features_df = pd.read_csv(selected_feature_path)
selected_features = selected_features_df["feature"].tolist()

# -------------------- Preprocessing --------------------
# Categorical and numerical columns from training data
categorical_cols = X.select_dtypes(include="object").columns.tolist()
numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Fill missing in train data
X[categorical_cols] = X[categorical_cols].fillna("Missing")
for col in numerical_cols:
    X[col] = X[col].fillna(X[col].median())

# Fill missing in test data
test_data[categorical_cols] = test_data[categorical_cols].fillna("Missing")
for col in numerical_cols:
    test_data[col] = test_data[col].fillna(test_data[col].median())

# Target Encoding
encoder = TargetEncoder(cols=categorical_cols)
X[categorical_cols] = encoder.fit_transform(X[categorical_cols], y)
test_data[categorical_cols] = encoder.transform(test_data[categorical_cols])

# Standard Scaling
scaler = StandardScaler()
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])
test_data[numerical_cols] = scaler.transform(test_data[numerical_cols])

# Final feature filtering
X_final = X[selected_features].copy()
X_test = test_data[selected_features].copy()

# -------------------- Train LightGBM on Full Dataset --------------------
lgb_train = lgb.Dataset(X_final, label=y, free_raw_data=False)

params = {
    'objective': 'binary',
    'metric': 'auc',
    'boosting_type': 'gbdt',
    'verbosity': -1,
    'num_leaves': 31,
    'learning_rate': 0.05,
    'feature_fraction': 0.9,
    'random_state': 42
}

model = lgb.train(
    params=params,
    train_set=lgb_train,
    num_boost_round=5000,
    callbacks=[
        lgb.log_evaluation(50)
    ]
)

model.save_model(model_path)
print(f"Model saved to {model_path}")

# -------------------- Predict on Test Set --------------------
y_test_pred = model.predict(X_test, num_iteration=model.best_iteration)
submission["isFraud"] = y_test_pred
submission.to_csv(submission_path, index=False)

print(f" Submission file saved to: {submission_path}")
```
